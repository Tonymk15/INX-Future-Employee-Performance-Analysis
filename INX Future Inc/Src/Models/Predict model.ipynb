Employee Performance Prediction

This notebook uses the Gradient Boosting classification model to predict employee performance ratings based on employee realted factors.Gradient Boosting model was selected beacause it achieved the highest performance during model evaluation with an accuracy 93.75%.The prediction uses employee information as input and generates a predicted performance rating.

**Import Libraries**

The required libraries are imported to load the trained model,scaler and feature information and also to prepare employee data for prediction.

```python
import pandas as pd
import numpy as np
import joblib
```

```python
model = joblib.load('gradient_boosting_model (3).pkl')
scaler = joblib.load('scaler (1).pkl')
model_features = joblib.load('model_features.pkl')

print("Model loaded successfully.")
print("Scaler loaded successfully.")
print("Number of model features:", len(model_features))
```

**Loading the employee Dataset**

The original employee performance dataset is loaded so as to provide the information required for making predictions.

```python
df = pd.read_excel("/content/INX_Future_Inc_Employee_Performance.xlsx")
```

```python
df.head()
```

```python
print("Dataset shape:", df.shape)
```

This is just to confirm that the prediction notebook is working with the same dataset used during model training

**Employee Input**

This is an example of an employee's profile using the same features that were used to train the model. The employee's information is processed in the same way as the training data before making a prediction.

```python
df.columns.tolist()
```

```python
employee = pd.DataFrame({
    'Age': [30],
    'Gender': ['Male'],
    'EducationBackground': ['Life Sciences'],
    'MaritalStatus': ['Single'],
    'EmpDepartment': ['Development'],
    'EmpJobRole': ['Developer'],
    'BusinessTravelFrequency': ['Travel_Rarely'],
    'DistanceFromHome': [5],
    'EmpEducationLevel': [3],
    'EmpEnvironmentSatisfaction': [4],
    'EmpHourlyRate': [75],
    'EmpJobInvolvement': [3],
    'EmpJobLevel': [2],
    'EmpJobSatisfaction': [4],
    'NumCompaniesWorked': [2],
    'OverTime': ['No'],
    'EmpLastSalaryHikePercent': [15],
    'EmpRelationshipSatisfaction': [3],
    'TotalWorkExperienceInYears': [6],
    'TrainingTimesLastYear': [3],
    'EmpWorkLifeBalance': [3],
    'ExperienceYearsAtThisCompany': [4],
    'ExperienceYearsInCurrentRole': [2],
    'YearsSinceLastPromotion': [1],
    'YearsWithCurrManager': [2],
    'Attrition': ['No']
})

employee
```

The employee's profile is using the original employee attributes.The input is then transformed using the same encoding and scaling procedures applied during model training.

**Prepare Employee Input**

The categorical variables in the employee input is converted into numerical values using one hot encoding. The same encoding approach used during the model training is applied to ensure that the input matches the format expected by the trained model.

```python
employee_encoded = pd.get_dummies(
    employee,
    drop_first=True
)

employee_encoded
```

```python
employee_encoded = employee_encoded.reindex(
    columns=model_features,
    fill_value=0
)

employee_encoded
```

```python
employee_scaled = scaler.transform(employee_encoded)
```

**Predict Employee Performance**

The prepared employee data is now passed to the trained Gradient Boosting model to predict the employee's performance rating.

```python
prediction = model.predict(employee_encoded)

print("Predicted Performance Rating:", prediction[0])
```
