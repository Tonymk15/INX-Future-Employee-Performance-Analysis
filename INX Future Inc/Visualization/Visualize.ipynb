INX Future Inc. Employee Performance Analysis

Data and Model Visualizations

This notebook contains key visualizations from the employee performance analysis and machine learning results.



**Import Libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**Load Employee Performance Data**

The employee performance dataset is loaded for visualization and analysis.

```python
df = pd.read_excel("/content/INX_Future_Inc_Employee_Performance.xlsx")
```

```python
df.shape
```

**Overall Employee Performance**

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='PerformanceRating'
)

plt.title('Distribution of Employee Performance Ratings')
plt.xlabel('Performance Rating')
plt.ylabel('Number of Employees')
plt.show()
```

The distribution shows the number of employees in each performance rating category. Performance Rating 3 represents the largest group of employees, indicating that it is the most common performance level in the dataset.

**Performance Rating by Department**

```python
department_performance = pd.crosstab(
    df['EmpDepartment'],
    df['PerformanceRating']
)

department_performance.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Employee Performance by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.legend(title='Performance Rating')
plt.tight_layout()
plt.show()
```

Employee performance varies across departments. The workforce is concentrated mainly in Sales, Development and Research & Development. Lower performance ratings are more noticeable in Sales and Research & Development suggesting that these departments may benefit from targeted performance improvement strategies.

**Machine Learning Model Performance**

```python
model_results = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Decision Tree',
        'Random Forest',
        'Gradient Boosting'
    ],
    'Accuracy': [
        0.8250,
        0.9083,
        0.9167,
        0.9375
    ],
    'F1-Score': [
        0.8177,
        0.9076,
        0.9128,
        0.9367
    ]
})

model_results
```

```python
model_results.plot(
    x='Model',
    y=['Accuracy', 'F1-Score'],
    kind='bar',
    figsize=(10, 6)
)

plt.title('Machine Learning Model Performance Comparison')
plt.xlabel('Model')
plt.ylabel('Score')
plt.xticks(rotation=20)
plt.ylim(0, 1)
plt.legend()
plt.tight_layout()
plt.show()
```

Gradient Boosting achieved the highest accuracy and F1-score among the four models evaluated. It achieved an accuracy of 93.75% and an F1-score of 93.67%, making it the selected model for employee performance prediction.

**Top Three Factors Affecting Employee Performance**

```python
top_features = pd.DataFrame({
    'Feature': [
        'Employee Environment Satisfaction',
        'Employee Last Salary Hike Percentage',
        'Years Since Last Promotion'
    ],
    'Importance': [
        0.276,
        0.245,
        0.196
    ]
})

top_features = top_features.sort_values(
    by='Importance',
    ascending=True
)

plt.figure(figsize=(10, 5))

plt.barh(
    top_features['Feature'],
    top_features['Importance']
)

plt.title('Top Three Important Features')
plt.xlabel('Feature Importance')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()
```

Employee Environment Satisfaction was the most influential feature in the Gradient Boosting model followed by Employee Last Salary Hike Percentage and Years Since Last Promotion. These factors should receive more attention when developing the strategies to support employee performance.
